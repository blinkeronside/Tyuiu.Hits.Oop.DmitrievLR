```
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}, \n Позиция: {Position}, \n Зарплата: {Salary}");
    }
}


public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"\n Отдел: {Department}");
    }
}


public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }

    
    private List<Employee> Team = new List<Employee>();

    public ProjectManager(string name, string position, double salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        Team.Add(this); // тимлидер
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    
    public string GenerateReport()
    {
        return $"Отчет по проекту: {ProjectName}";
    }

    
    public List<string> GetTeamMembers()
    {
        List<string> names = new List<string>();
        

        foreach (var member in Team)
        {
            names.Add(member.Name);
        }
        return names;
    }

    
    public void AddToTeam(Employee employee)
    {
        Team.Add(employee);
    }
}


public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
}



var employee = new Employee("Иван", "менеджер", 80000);
var manager = new Manager("Петр", "Программист", 120000, "Разработка");
var projectManager = new ProjectManager("Анна", "Проектный менеджер", 150000, "R&D", "Проект X");

projectManager.AddToTeam(employee);
projectManager.AddToTeam(manager);

employee.PrintInfo();
Console.WriteLine();

manager.PrintInfo();
Console.WriteLine();

projectManager.PrintInfo();
Console.WriteLine();

Console.WriteLine("Отчет: " + projectManager.GenerateReport());
Console.WriteLine("Члены команды: " + string.Join(", ", projectManager.GetTeamMembers()));
```
